```python
import pandas as pd
import numpy as np

df = pd.read_csv("Social_Network_Ads.csv")
```

```python
x = df.drop(columns=["Purchased"])
y = df["Purchased"]
x  , y
```

```python
x.drop(columns=["User ID"] , inplace=True)
```

```python
x.head(10)
```

```python
x.isna().sum()
sex = pd.get_dummies(x["Gender"] , drop_first=True , dtype="int64")
df
sex
x["sex"] = sex
x.drop(columns=["Gender"] , inplace=True)
x["sex"]
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
ss.fit(x)

scaled_features = ss.transform(x)
scaled_features
```

```python
scale = pd.DataFrame(scaled_features, columns=x.columns)
scale["Purchased"] = y
scale
```

```python
scale.drop("Purchased" , axis=1)
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(scale.drop("Purchased" , axis=1) , scale["Purchased"] , test_size = 0.2 , random_state = 42)
x_train.shape , x_test.shape , y_train.shape , y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(x_train , y_train)
```

```python
pred = lr.predict(x_test)
```

```python
pred
```

```python
from sklearn.metrics import classification_report , confusion_matrix , accuracy_score

print(classification_report(y_test , pred))
print(confusion_matrix(y_test , pred))
print(accuracy_score(y_test , pred))
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train , y_train)
pred = knn.predict(x_test)
print(pred)
print(y_test)
print(classification_report(y_test , pred))
print(confusion_matrix(y_test , pred))
print(accuracy_score(y_test , pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

dtc = DecisionTreeClassifier()
dtc.fit(x_train , y_train)
pred = dtc.predict(x_test)
print(pred)
print(y_test)
print(classification_report(y_test , pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(n_estimators = 225 , random_state=42)
rfc.fit(x_train , y_train)
pred = rfc.predict(x_test)
print(pred)
print(y_test)
print(classification_report(y_test , pred))
```

```python
from sklearn.svm import SVC

svc = SVC()
svc.fit(x_train , y_train)
pred = svc.predict(x_test)
print(pred)
print(y_test)
print(classification_report(y_test , pred))
```

```python
from sklearn.ensemble import GradientBoostingClassifier

gfc = GradientBoostingClassifier()
gfc.fit(x_train , y_train)
pred = gfc.predict(x_test)
print(pred)
print(y_test)
print(classification_report(y_test , pred))
```

```python
from sklearn.ensemble import AdaBoostClassifier

ada = AdaBoostClassifier()
ada.fit(x_train , y_train)
pred = ada.predict(x_test)
print(pred)
print(y_test)
print(classification_report(y_test , pred))
```
